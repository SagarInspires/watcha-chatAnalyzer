```python
import re
import pandas as pd
```

```python
f = open('WhatsApp Chat with Ansh Prem.txt','r',encoding='utf-8')
```

```python
data=f.read()
```

```python
print(data)
```

```python
#convert these rows into table of various cols
#use regex
pattern ='\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2}\s-\s'

```

```python
messages = re.split(pattern,data)[1:]
messages    #now we have put all messages into a list
```

```python
dates = re.findall(pattern,data)
dates
```

```python
#now we've two lists, date time,message string, now make a df using two lists as columns,mapped correspondingly
df = pd.DataFrame({'user_message':messages,'message_date':dates})
#converting message_date type, difft format
df['message_date'] = pd.to_datetime(df['message_date'],format='%d/%m/%Y, %H:%M - ')
df.rename(columns={'message_date':'date'},inplace=True)
df.tail()
```

```python
#total message count
df.shape
```

```python
#now separate users and messages
users =[]
messages =[]
for message in df['user_message']:
    entry = re.split('([\w\W]+?):\s',message) #split means separate over given pattern
    if entry[1:]: #user name
        users.append(entry[1])
        messages.append(entry[2])
    else:            #when there is no col means mark it as group notification
        users.append('group_notification')
        messages.append(entry[0])
df['user'] = users
df['message'] =messages
df.drop(columns=['user_message'],inplace=True)
df.head()
```

```python
df['year']=df['date'].dt.year #making a year col
```

```python
df.head()
```

```python
#extracting month,day,hour,min,sec (always recorded 00) from date column
df['month']=df['date'].dt.month_name()
```

```python
df['day']=df['date'].dt.day
```

```python
df['hour'] = df['date'].dt.hour
```

```python
df['minute'] =df['date'].dt.minute
```

```python
#now for number of words enter in message col
words=[]
for message in df['message']: #all messages got split
    words.extend(message.split())
```
